```python
from scipy.stats import pearsonr
```

```python
def calculate_correlation(input_table, trial_store, metric):
    
    trial_data = input_table[input_table["STORE_NBR"] == trial_store]

    correlations = []

    for store in input_table["STORE_NBR"].unique():

        if store != trial_store:

            control_data = input_table[input_table["STORE_NBR"] == store]

            merged = trial_data.merge(
                control_data,
                on="YEAR_MONTH",
                suffixes=("_trial", "_control")
            )

            if len(merged) > 1:

                corr, _ = pearsonr(
                    merged[f"{metric}_trial"],
                    merged[f"{metric}_control"]
                )

                correlations.append([store, corr])

    correlation_df = pd.DataFrame(
        correlations,
        columns=["Control_Store", "Correlation"]
    )

    return correlation_df.sort_values(
        by="Correlation",
        ascending=False
    )
```

```python
def calculate_magnitude(input_table, trial_store, metric):
    trial_data = input_table[input_table["STORE_NBR"] == trial_store]
    magnitudes = []
    
    for store in input_table["STORE_NBR"].unique():
        if store != trial_store:
            control_data = input_table[input_table["STORE_NBR"] == store]
            merged = trial_data.merge(control_data, on="YEAR_MONTH", suffixes=("_trial", "_control"))
            
            # CRITICAL FIX: Only consider stores with full 7-month data overlap
            # This completely removes the false '1.0' perfect correlations from empty stores (like store 31)!
            if len(merged) == 7: 
                diff = abs(merged[f"{metric}_trial"] - merged[f"{metric}_control"])
                normalized_dist = 1 - (diff / merged[f"{metric}_trial"])
                mean_magnitude = normalized_dist.mean()
                magnitudes.append([store, mean_magnitude])
                
    magnitude_df = pd.DataFrame(magnitudes, columns=["Control_Store", "Magnitude"])
    return magnitude_df.sort_values(by="Magnitude", ascending=False)

def find_best_control(input_table, trial_store):
    # Get Sales Correlation and Magnitude
    corr_sales = calculate_correlation(input_table, trial_store, "TOT_SALES").rename(columns={"Correlation": "Corr_Sales"})
    mag_sales = calculate_magnitude(input_table, trial_store, "TOT_SALES").rename(columns={"Magnitude": "Mag_Sales"})
    
    # Get Customer Correlation and Magnitude
    corr_cust = calculate_correlation(input_table, trial_store, "TOTAL_CUSTOMERS").rename(columns={"Correlation": "Corr_Cust"})
    mag_cust = calculate_magnitude(input_table, trial_store, "TOTAL_CUSTOMERS").rename(columns={"Magnitude": "Mag_Cust"})
    
    # Combine all scores
    scores = corr_sales.merge(mag_sales, on="Control_Store")\
                       .merge(corr_cust, on="Control_Store")\
                       .merge(mag_cust, on="Control_Store")
    
    # Compute the final averaged score across trends (correlation) and size (magnitude)
    scores["Combined_Score"] = (scores["Corr_Sales"] + scores["Mag_Sales"] + scores["Corr_Cust"] + scores["Mag_Cust"]) / 4
    return scores.sort_values(by="Combined_Score", ascending=False)
```
